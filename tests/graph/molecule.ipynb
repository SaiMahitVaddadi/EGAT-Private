```python
import sys
sys.path.append("/Users/svaddadi/Documents/GitHub/")
import EGAT
from EGAT import src
from EGAT.src.graph.molecular.Molecule import MoleculeFeaturizer
```

```python
from EGAT.src.params.graph import GraphParams

params = GraphParams()
def printattributes(atr):
    attributes = [attr for attr in dir(atr) if not callable(getattr(atr, attr)) and not attr.startswith("__")]
    for attr in attributes:
        value = getattr(atr, attr)
        print(f"{attr}: {type(value)} = {value}")

    print ('There are {} attributes in the MolMatDesc class!'.format(len(attributes)))

printattributes(params)
```

```python

graph = MoleculeFeaturizer('CCCCC',params)
printattributes(graph)
graph.run()
printattributes(graph)
```
