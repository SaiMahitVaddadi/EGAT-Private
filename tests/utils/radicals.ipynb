```python
import sys
sys.path.append("/Users/svaddadi/Documents/GitHub/")
import EGAT
from EGAT import src
from EGAT.src.utils.descriptors.egat.molmatdesc import MolMatDesc
from EGAT.src.utils.descriptors.egat.radicals import RDKElectronInfo,YARPElectronInfo
from rdkit import Chem 
```

```python
sample = '[CH2]C1=CC=CC=C1'
descriptor = MolMatDesc(sample)
descriptor.run()
descriptor.new_mol
```

```python
rdkit = RDKElectronInfo(descriptor)
rdkit.run()

attributes = [attr for attr in dir(rdkit) if not callable(getattr(rdkit, attr)) and not attr.startswith("__")]
for attr in attributes:
    value = getattr(rdkit, attr)
    print(f"{attr}: {type(value)} = {value}")

print ('There are {} attributes in the MolMatDesc class!'.format(len(attributes)))
```
