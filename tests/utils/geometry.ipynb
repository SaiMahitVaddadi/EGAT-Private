```python

attributes = [attr for attr in dir(geom) if not callable(getattr(geom, attr)) and not attr.startswith("__")]
for attr in attributes:
    value = getattr(geom, attr)
    print(f"{attr}: {type(value)} = {value}")

print ('There are {} attributes in the MolMatDesc class!'.format(len(attributes)))
```
